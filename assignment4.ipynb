```python
# q2.
 
def factorial(n):
    if n == 0 or n == 1:
        return 1 
    
    return n * factorial(n-1)

num = 10
print("factorial  of the number ",num , "is  : ", factorial(num))
```

```python
# q4.

import json
student = {
    "name": "nikhil",
    "age": 22,
    "course": "B-tech",
    "city": "Jaipur"
}

with open("student.json", "w") as file:
    json.dump(student, file, indent=4)

print("Dictionary stored in student.json")

with open("student.json", "r") as file:
    data = json.load(file)

print("\nData read from JSON file:")
print(data)
```
