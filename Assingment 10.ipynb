```python
# 1.
# K-Nearest Neighbors (KNN) is a supervised machine learning algorithm used for both classification and regression problems.

# It predicts the output of a new data point based on the K nearest data points in the training dataset.

# Working Process of KNN
# Choose the value of K.
# Calculate the distance between the new data point and all training points.
# Common distance metric: Euclidean Distance
```

```python
# 2.
# The value of K determines how many nearest neighbors participate in prediction.

# K Too Small (e.g., K=1)
# Sensitive to noise.
# Can overfit.
# High variance.

# Example:
# A single incorrect neighbor may lead to wrong prediction.

# K Too Large
# May include unrelated points.
# Can underfit.
# High bias.

# Example:
# If K equals almost the whole dataset, local patterns are ignored.
```

```python
#3.

# Feature	KNN Classification	KNN Regression
# Output	Category/Class	Continuous Value
# Example	Pass/Fail	Salary Prediction
# Prediction Method	Majority Voting	Average of Neighbors
# Evaluation	Accuracy, Precision	MAE, MSE, RMSE

# Advantages of KNN

# Easy to understand.
# No training phase.
# Works well for small datasets.
# Can handle multiclass problems.

# Disadvantages of KNN

# Slow for large datasets.
# Sensitive to irrelevant features.
# Requires feature scaling.
# Memory intensive.
```

```python
# 4.

import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score


data = {
    'Study_Hours':[1,2,3,4,5,6],
    'Attendance':[50,55,60,70,75,85],
    'Result':['Fail','Fail','Fail','Pass','Pass','Pass']
}

df = pd.DataFrame(data)

print(df)

encoder = LabelEncoder()

df['Result'] = encoder.fit_transform(df['Result'])

print(df,"\n\n")

X = df[['Study_Hours','Attendance']]
y = df['Result']

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.33,
    random_state=42
)
model = KNeighborsClassifier(n_neighbors=3)

model.fit(X_train,y_train)

prediction = model.predict([[4.5,72]])

result = encoder.inverse_transform(prediction)

print("Prediction:",result[0])


y_pred = model.predict(X_test)

acc = accuracy_score(y_test,y_pred)

print("Accuracy:",acc)
```

```python
# 6

import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix,classification_report

data = {
    'Weight':[120,130,140,170,180,200],
    'Size':[6,6.5,7,8,8.5,9],
    'Fruit':['Apple','Apple','Apple','Orange','Orange','Orange']
}

df = pd.DataFrame(data)

encoder = LabelEncoder()
df['Fruit'] = encoder.fit_transform(df['Fruit'])

X = df[['Weight','Size']]
y = df['Fruit']

X_train,X_test,y_train,y_test = train_test_split(
    X,y,test_size=0.33,random_state=42
)

model = KNeighborsClassifier(n_neighbors=3)

model.fit(X_train,y_train)

prediction = model.predict([[160,7.5]])

print("Predicted Fruit:",
      encoder.inverse_transform(prediction)[0])

y_pred = model.predict(X_test)

print("Confusion Matrix")
print(confusion_matrix(y_test,y_pred))

print("\nClassification Report")
print(classification_report(y_test,y_pred))
```
