```python
# q1.

import math

num = int(input("enter the num :"))
temp = num
sumfact = 0

while temp > 0:
    digit = temp % 10
    sumfact += math.factorial(digit)
    temp //= 10

if sumfact == num:
    print("peterson number")
else:
    print("not a peterson number ")

```

```python
# q2 

def prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

num = input("Enter a number: ")
flag = True

for i in range(len(num)):
    rotation = int(num[i:] + num[:i])
    if not prime(rotation):
        flag = False
        break

if flag:
    print("Circular Prime")
else:
    print("Not a Circular Prime")
```

```python
#q3. harshad number

num = int(input("Enter a number: "))

digit_sum = sum(int(d) for d in str(num))

if num % digit_sum == 0:
    print("Harshad Number")
else:
    print("Not a Harshad Number")
```

```python
# q4.

num = int(input("Enter a number: "))

while num > 9:
    s = 0
    while num > 0:
        s += num % 10
        num //= 10
    num = s

if num == 1:
    print("Magic Number")
else:
    print("Not a Magic Number")
```

```python
# q5.

n = int(input("Enter number of rows: "))

for i in range(n):
    num = 1
    for j in range(i + 1):
        print(num, end=" ")
        num = num * (i - j) // (j + 1)
    print()
```
