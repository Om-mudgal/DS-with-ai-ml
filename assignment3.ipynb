```python
# q1.

def calculate_bill(price,quantity,discount):
    total_bil = (price * quantity)
    discount_bil = (total_bil*discount)/100
    final =  total_bil-discount_bil

    return final

# price = 500
# quantity = 3
# discount = 10

price = int(input("enter the price : "))
quantity = int(input("enter the quantity : "))
discount = int(input("enter the discount without % : "))

payment = calculate_bill(price,quantity,discount)

print("payment of the final : ",payment)

```

```python
# q2.

def calculate_salary (basic,bonus=2000, tex=10):
    basicsalary = (basic+bonus)
    taxpercentage = basicsalary - (basicsalary *tex)/100

    netsalary = taxpercentage

    return netsalary

basic = int(input("enter the basic :"))

final1 = calculate_salary (basic,bonus=2000, tex=10)

print("final net basic salary : ",final1)

final2 = calculate_salary (30000,2000,15)

print("final net salary all argumant : ", final2)
```

```python
#q3

def find_max(*number):
    larges = number[0]

    for num in number:
       if num > larges:
           larges = num

    return larges

result = find_max(10,45,23,67,12)

print(result)
```
