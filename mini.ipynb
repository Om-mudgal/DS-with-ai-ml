```python

import re

password = input("Enter your password: ")
print("your password : ",password)
length = len(password) >= 8
upper = re.search(r"[A-Z]", password)
lower = re.search(r"[a-z]", password)
digit = re.search(r"[0-9]", password)
special = re.search(r"[@#$%^&*()_+\-=\[\]{};':\"\\|,.<>/?]", password)

score = 0

if length:
    score += 1
if upper:
    score += 1
if lower:
    score += 1
if digit:
    score += 1
if special:
    score += 1

if score <= 2:
    strength = "Weak"
elif score <= 4:
    strength = "Medium"
else:
    strength = "Strong"

print("\nPassword Strength:", strength)


if not length:    # find the feedback
    print("- Password should be at least 8 characters long.")
if not upper:
    print("- Add at least one uppercase letter.")
if not lower:
    print("- Add at least one lowercase letter.")
if not digit: 
    print("- Add at least one number.")
if not special:
    print("- Add at least one special character.")

if strength == "Strong":
    print("- Excellent! Your password is secure.")
```

```python
import json
import os

FILE_NAME = "tasks.txt"

def load_tasks():
    if os.path.exists(FILE_NAME):
        with open(FILE_NAME, "r") as f:
            return json.load(f)
    return []

def save_tasks(tasks):
    with open(FILE_NAME, "w") as f:
        json.dump(tasks, f)

def add_task(tasks):
    name = input("Task Name: ")
    priority = input("Priority (High/Medium/Low): ")
    tasks.append({
        "name": name,
        "status": "Pending",
        "priority": priority
    })
    save_tasks(tasks)
    print("Task Added Successfully.")

def view_tasks(tasks):
    if not tasks:
        print("No Tasks Found.")
        return

    for i, task in enumerate(tasks, start=1):
        print(f"{i}. {task['name']} | {task['status']} | {task['priority']}")

def delete_task(tasks):
    view_tasks(tasks)
    try:
        index = int(input("Enter Task Number to Delete: ")) - 1
        tasks.pop(index)
        save_tasks(tasks)
        print("Task Deleted.")
    except:
        print("Invalid Choice")

def mark_completed(tasks):
    view_tasks(tasks)
    try:
        index = int(input("Enter Task Number: ")) - 1
        tasks[index]["status"] = "Completed"
        save_tasks(tasks)
        print("Task Marked Completed.")
    except:
        print("Invalid Choice")

def search_task(tasks):
    keyword = input("Enter Keyword: ").lower()
    found = False

    for task in tasks:
        if keyword in task["name"].lower():
            print(task)
            found = True

    if not found:
        print("No Matching Task Found.")

def statistics(tasks):
    total = len(tasks)
    completed = sum(1 for t in tasks if t["status"] == "Completed")
    pending = total - completed

    print("\\nTask Statistics")
    print("Total:", total)
    print("Completed:", completed)
    print("Pending:", pending)

tasks = load_tasks()

while True:
    print("\\n===== TO-DO LIST MENU =====")
    print("1. Add Task")
    print("2. View Tasks")
    print("3. Delete Task")
    print("4. Mark Task Completed")
    print("5. Search Task")
    print("6. Task Statistics")
    print("7. Exit")

    choice = input("Enter Choice: ")

    if choice == "1":
        add_task(tasks)
    elif choice == "2":
        view_tasks(tasks)
    elif choice == "3":
        delete_task(tasks)
    elif choice == "4":
        mark_completed(tasks)
    elif choice == "5":2

        search_task(tasks)
    elif choice == "6":
        statistics(tasks)
    elif choice == "7":
        print("Exiting Application...")
            break
    else:
        print("Invalid Choice")
```
