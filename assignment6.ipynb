```python
# q1.

class Emp:
    def __init__(self,emp_name,emp_id,salary):
        self.emp_name = emp_name
        self.emp_id = emp_id
        self.salary = salary
    
    def display(self):
        print("name :",self.emp_name)
        print("emp_id",self. emp_id)
        print("emp_salary",self.salary)

    e1 = Emp("akhilesh",407,50000)
    e4 = Emp("Om", 101, 50000)
    e2 = Emp("kushagra", 102, 60000)
    e3 = Emp("nikhil", 103, 55000)

    e1.display()
    e2.display()
    e3.display()
    e4.display()

```

```python
# q2 

class Calculator:
    def add(self, a, b):
        return a + b
    def sub(self, a, b):
        return a-b
    def mul (self, a, b):
        return a *b
    def div(self, a, b):
        if b !=0:
            return a/b
        return ("divided by zero")
    
c1 = Calculator()

print("adition of the a and b ", c1.add(2,5))
print("subtraction of the a and b ", c1.sub(2,5))
print("multiplication of the a and b ", c1.mul(2,5))
print("division of the a and b ", c1.div(2,5))


```
